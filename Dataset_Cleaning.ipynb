# Práctica 14: Normalización de los nombres de la columna.

* Normaliza los nombres de las columnas, los cuales deberán estar en minúsculas.
* Asimismo, reemplaza los espacios con guion bajo
* Imprime los nombres de las columnas para demostrar que has realizado la normalización.

```python
import pandas as pd
from IPython.display import display, Markdown

# First, load the data and ensure the first column is named correctly
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/miaad-nyc-r-s.csv', skipfooter=1, engine='python')

# Set the first column name to 'INDEX MIAAD' as we did before
# we need to do it again since we did not save the dataset
df.rename(columns={df.columns[0]: 'INDEX MIAAD'}, inplace=True)

display(Markdown("### Original column names:"))
#display(df.iloc[:, :5].columns.tolist())
display(df.columns.tolist())
print()

# Step 1 & 2: Normalize column names (lowercase and replace spaces with underscores)
# First of all, we need to create a list of new column names
normalized_columns = []
display(Markdown("### Normalizing..."))
for col in df.columns:
    new_col = col.lower().replace(' ', '_') # Convert to lowercase and replace spaces with underscores
    normalized_columns.append(new_col) # append modify column to the list of new columns
    print(f"'{col}' -> '{new_col}'") #old column , new column

"""
Assigning a list of normalized names to df.columns in pandas (df.columns = normalized_columns) is possible because
the .columns attribute of a pandas DataFrame is designed to be a mutable, index-like object that holds column label
"""
df.columns = normalized_columns # Apply the new column names. columns is an Index. Done.

display(Markdown("### Variable names after normalization:"))
display(df.columns.tolist())
print()

# Optional: Show first few rows to demonstrate the new column names in action
display(Markdown("### First 2 rows with normalized column names:"))
display(df.iloc[:, :5].head(3))
```

* Escribe qué tipo de dato tienen asignado. Si no son correctos (flotante), realiza el cambio.

```python
# Load and prepare
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/ClassFiles/miaad-nyc-r-s.csv', skipfooter=1, engine='python') #skip last bad row
df.rename(columns={df.columns[0]: 'INDEX MIAAD'}, inplace=True) #chage name of first column
df.columns = [col.lower().replace(' ', '_') for col in df.columns] #make underscore variable names

# Show original types
display(Markdown("### Original:"))
display( df.dtypes)

# Columns to convert
float_cols = ['sale_price', 'land_square_feet', 'gross_square_feet']

# list needs_conversion will keep what needs to be converted as a dictionary of string variable names.
"""
    This creates a mapping dictionary (needs_conversion) of column names to their current data types,
    but only for columns listed in float_cols that exist in the DataFrame (df.columns) and are not already float64 or float32. It identifies columns requiring conversion to float.
"""

"""
A dictionary in action in one line loop.
col: df[col].dtype: The dictionary key is the column name (col),
and the value is its current pandas data type (df[col].dtype).
"""
needs_conversion = {col: df[col].dtype for col in float_cols # before next loop, keeps track of column name saving it in needs_conversion

                   if col in df.columns and df[col].dtype not in ['float64', 'float32']}

# printing current assign values to the columns into question
display(Markdown("###  Current  assigned values to the columns in question:"))
for key, value in needs_conversion.items():
    print(f"Column: {key}, Type: {value}")


if needs_conversion:
    for col in needs_conversion: #gets the key (col) string variable name to index the dataframe df
        df[col] = pd.to_numeric(df[col], errors='coerce').astype('float64') ## indexing column to be change, then changes type
        print(f"\nConverting...{col} to float64")

# Show new column types
print()
display(Markdown("### Using dtypes to see conversion on specific columns:\n"))
display(df.dtypes[float_cols])
# display all columns
# Show original types
display(Markdown("### Modified DataFrame column types:"))
display( df.dtypes)
```

# Práctica 16:
* Genera el diccionario de datos del dataframe preprocesado.
* Imprime el contenido de dicho diccionario

```python
import pandas as pd

# Load and prepare the data
df = pd.read_csv('/content/drive/MyDrive/ClassFiles/miaad-nyc-r-s.csv', skipfooter=1, engine='python') #skip last row
df.rename(columns={df.columns[0]: 'INDEX MIAAD'}, inplace=True) # rename first colum (variable name)
df.columns = [col.lower().replace(' ', '_') for col in df.columns] # normalize column's names

""" Make sure again the floating columns are written correctly,
    change them again using a loop;
    it's easier now since you don't need to check them and print those needed.
"""

float_cols = ['sale_price', 'land_square_feet', 'gross_square_feet']
for col in float_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce').astype('float64')

# 1. Generate a simple data dictionary from preprocessed dataframe
# Store the datadictionary using a Dictionary
data_dict = {}
"""
the key is going to be the column's name, and its values the column's associated information,
so each pair becomes an entry in the final dataset Dictionary
"""
# its is easier to do, just iterate through the dataframe columns,
## then build an entry of the dictionary
# insert that entry into the data_dict dictionary
#all the valid information to build the dictionary entries are in each variable column
# just use pandas dataframe built-in functions to ask for
for col in df.columns:
    data_dict[col] = {
        'dtype': df[col].dtype, #data type of the variable column
        'non_null_count': df[col].count(), # how much not null values has that variable column
        'null_count': df[col].isnull().sum(), #to count null values in current column
        'null_percentage': round(df[col].isnull().sum() / len(df) * 100, 2), # to save percentage of null values in current column
        'unique_values': df[col].nunique() # how many different variable's values can be or has been assigned
    }

    # Add sample values for context, sample values are possible values taken from current ones
    # it adds a sample of 3 current values : min (3,...) or [] if not sample values to keep
    sample_values = df[col].dropna().sample(min(3, df[col].count())).tolist() if df[col].count() > 0 else []
    data_dict[col]['samples'] = sample_values

    # Add numeric stats for numerical variable columns
    # new columns min, max, meann, std are added on the fly
    # once again, information is gotten from built-in functions of pandas dataframe
    if pd.api.types.is_numeric_dtype(df[col]):
        data_dict[col]['min'] = df[col].min()
        data_dict[col]['max'] = df[col].max()
        data_dict[col]['mean'] = round(df[col].mean(), 2)
        data_dict[col]['std'] = round(df[col].std(), 2)

# 2. Print the contents of the Dictionary in a readable way
print("=" * 60)
display(Markdown("###  DATA DICTIONARY"))
display(Markdown("*Variable names are changed to uppercase*"))
print("=" * 60)


for col_name, col_info in data_dict.items(): # iterate through the dictionary list
    print(f"\n {col_name.upper()}")
    print("-" * 40)
    for key, value in col_info.items(): # each element is itself a dictionary
        print(f"  {key}: {value}")
```
